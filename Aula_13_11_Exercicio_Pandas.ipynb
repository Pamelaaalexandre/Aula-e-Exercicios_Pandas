```python
import pandas as pd
```

```python
df_vendas_geek_store = pd.read_csv("/content/vendas_geek_store.csv")
display(df_vendas_geek_store)
```

```python
# LUCRO UNIDADE E LUCRO FINAL
df_vendas_geek_store["Receita"] = df_vendas_geek_store["Preco_Venda"] * df_vendas_geek_store["Quantidade_Vendida"]
df_vendas_geek_store["Lucro_Unitario"] = df_vendas_geek_store["Preco_Venda"] - df_vendas_geek_store["Custo_Fabricante"]
df_vendas_geek_store["Lucro_Final"] = df_vendas_geek_store["Lucro_Unitario"] * df_vendas_geek_store["Quantidade_Vendida"]
display(df_vendas_geek_store[["Produto", "Categoria", "Receita", "Lucro_Unitario", "Lucro_Final"]])
```
